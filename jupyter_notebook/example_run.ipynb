```python
import time
import h5py

import numpy as np
import matplotlib.pyplot as plt
import sys

from chemrixs.average import Average
from chemrixs.smalldata import SmallData
from chemrixs.process import Reduced
from chemrixs.utils import *
from pathlib import Path
%matplotlib widget 
```

```python
# Define base folder and folder where processed data is supposed to be stored
base_folder =  '/sdf/data/lcls/ds/rix/'
proc_folder = './proc/'
#experiment number
exp = 'rix101331225'

# Define Dark run for background and runs to analyse
bgrun = 106
runs=[112,113,114,116,117,120,121,122,123]

# Background small data file
bgname = f'{base_folder}/{exp}/hdf5/smalldata/{exp}_Run{bgrun:04d}.h5'

# yaml files containing input parameters for calibration, which detectors to use etc
######## IMPORTANT! CHECK THESE FILES! #############################################
fyaml = '../roi_input.yml'
bgyaml = '../BGroi_input.yml'


# Run processing on unprocessed runs and do averaging
avg = Average(runs, proc_path = 'proc/Run', avg = 'mean', output_path = f'avg/{runs[0]}_to_{runs[-1]}.h5',
             raw_path = f'{base_folder}/{exp}/hdf5/smalldata/{exp}_Run' , bgpath = bgname, fyaml=fyaml, bgyaml=bgyaml, 
             save= True,norm = True)

avg.plot_svls2D(figsize=(12,4),savefig=True)
avg.plot_svls1D(figsize=(12,4))
```
