```python
base_folder =  '/sdf/data/lcls/ds/rix/'
proc_folder = './proc/'

run=100
exp = 'rix101331225'


fname = f'{base_folder}/{exp}/hdf5/smalldata/{exp}_Run{run:04d}.h5'
with open('../roi_input.yml', 'r') as file:
    conf = yaml.safe_load(file)

bigdata = h5py.File(fname,'r')
```

```python
with  h5py.File('testfile_Run0000.h5','w') as testfile:
    for detector in conf['int_detectors']:
        attrdict=conf['int_detectors'][detector]['attrdict']
        for attr in conf[attrdict]:
            print(conf[attrdict][attr])
            if attr=='full_area':
                try:
                    dat =np.asarray(bigdata['intg'][detector][conf[attrdict][attr]][:10,:564])
                    print(f'intg/{detector}/{conf[attrdict][attr]}')
                    testfile.create_dataset(f'intg/{detector}/{conf[attrdict][attr]}', data=dat)
                except:
                    print(f'boop')

            elif np.logical_or(np.logical_or((attr=='fim_0'),(attr=='fim_1')),(attr=='apds')):
                try:
                    dat =np.asarray(bigdata['intg'][detector][conf[attrdict][attr]][:10,:,:100])
                    print(f'intg/{detector}/{conf[attrdict][attr]}')
                    testfile.create_dataset(f'intg/{detector}/{conf[attrdict][attr]}', data=dat)
                except:
                    print(f'boop')

            else:
                print(f'intg/{detector}/{conf[attrdict][attr]}')
                try:
                    dat =np.asarray(bigdata['intg'][detector][conf[attrdict][attr]][:10])
                    testfile.create_dataset(f'intg/{detector}/{conf[attrdict][attr]}', data=dat)
                except:
                    print(f'boop')
    
    # testfile.create_dataset(f'{conf['epics']['key']}', data=dat)
    for key in conf['epics']['attr']:
        attr = conf['epics']['attr'][key]
        epics_key = conf['epics']['key']
        dat=np.asarray(bigdata[epics_key][attr][:10])
        testfile.create_dataset(f'{epics_key}/{attr}', data=dat)
            # FIXME
            # if dataset does not exist:
            #     testfile.create_dataset(f'intg/{detector}/{conf[attrdict][attr]}',1,dtype='i', data=np.asarray(dat))
            # else:
            #     delete datase
            #     testfile.create_dataset(f'intg/{detector}/{conf[attrdict][attr]}',1,dtype='i', data=np.asarray(dat))
            
```
